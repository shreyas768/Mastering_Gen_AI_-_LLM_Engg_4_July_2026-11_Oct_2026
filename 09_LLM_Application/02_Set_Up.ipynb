```python
%%writefile app.py

import streamlit as st
from openai import OpenAI


st.set_page_config(page_title="26 Sep chatbot")
st.title('Demo Chatbot')

st.subheader('Your Personal Information', divider='rainbow')
name = st.text_input(label='Your Name', max_chars=40, placeholder="Enter your name here")
st.write(f"Your Name: {name}")
experience = st.text_area(label='Experience', max_chars=None, placeholder="Enter your experience here")
st.write(f"Experience: {experience}")
skills = st.text_input(label='Skills', max_chars=40, placeholder="Enter your relevant skills here")
st.write(f"Skills: {skills}")

st.subheader("Company and Position")

col1, col2 = st.columns(2)

with col1:
  level = st.radio("Choose your level:",
                   key = "visibility",
                   options = ["Junior","Intermediate","Senior"])

with col2:
  position = st.selectbox("Choose your position:",
                   options = ["ML Engineer","Gen AI Engineer","Data Scientist","FInancial Analyst"])

company = st.selectbox("Choose your Company:",
                   options = ["Amazon","Meta","LinkedIn","Nestle","Uber"])



st.write(f"Your Information: {level} {position} at {company}")

client = OpenAI(api_key = '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')


if "openai_model" not in st.session_state:
  st.session_state['openai_model'] = 'gpt-4o'

if 'messages' not in st.session_state:
  st.session_state.messages = [{'role':'system',
                                'content':f'You are a HR executive that interviews an interviewee called {name} with experience {experience} and skills {skills}. You have to interview him/her for {level} {position} postion at the company {company}'}]


for message in st.session_state.messages:
  if message['role'] != "system":
    with st.chat_message(message['role']):
      st.markdown(message['content'])

if prompt := st.chat_input("Your Answer"):
  st.session_state.messages.append({'role':'user',
                                    'content':prompt})
  with st.chat_message('user'):
    st.markdown(prompt)

  with st.chat_message('assistant'):
    stream = client.chat.completions.create(
        model = st.session_state['openai_model'],
        messages = [
            {'role':m['role'], 'content':m['content']} for m in st.session_state.messages
        ],
        stream = True
    )

    response = st.write_stream(stream)
  st.session_state.messages.append({'role':'assistant',
                          'content':response})

```
