```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0,10)
y = 2*x
```

```python
fig = plt.figure()

ax  = fig.add_axes([0,0,1,1])

ax.plot(x,x**2,'bo',label='x squared',marker='+',ms=10)
ax.plot(x,x**3, label = 'x cubed',ls='--',lw=5,marker='D',mew=5,mec='green',mfc='yellow',ms=16,fillstyle='bottom')
ax.legend(fontsize='x-small',loc=0)

plt.show()
```
