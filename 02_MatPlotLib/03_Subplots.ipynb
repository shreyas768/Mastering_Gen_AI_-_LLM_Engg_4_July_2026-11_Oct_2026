```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.linspace(0,10,11)
b = a**4
```

```python
x = np.arange(0,10)
y = 2*x
```

```python
fig,axes = plt.subplots()

axes.plot(x,y)

plt.show()
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

axes[0].plot(a,b)

axes[1].plot(b,a)
axes[1].set_title('this is 1st figure')


fig.suptitle('super title',fontsize=20)

plt.show()
```

```python

fig,axes = plt.subplots(nrows=3,ncols=2,figsize=(12,8))

axes[0,0].set_xlabel('This is x label for this figure')

axes[0,1].plot(x,y)

axes[1,0].plot(a,b)
axes[1,0].set_title('This is title')

axes[-2,-1].plot(y,x)

fig.subplots_adjust(wspace=0.9,hspace=0.5)

plt.show()
```
