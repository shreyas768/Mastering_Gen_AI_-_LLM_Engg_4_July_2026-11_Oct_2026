```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(10)
x
```

```python
y = 2*x
y
```

```python
plt.plot(x,y)
plt.xlabel("Put X axis title here")
plt.ylabel("Put y axis title here")
plt.title('Image Title here')
plt.xlim(2,6)
plt.ylim(2.5,12.5)
plt.show() #same as print command
```
