```python
import nltk

nltk.download('stopwords')

from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words('english')
```

```python
print(en_stopwords)
```

```python
phrase = "Once upon a time, there was course on Gen AI liked by everyone"
```

```python
phrase.split()
```

```python
' '.join([w for w in phrase.split() if w not in en_stopwords])
```

```python
import string

translator = str.maketrans('','',string.punctuation)

clean_phrase = phrase.translate(translator)
```

## str.maketrans(x,y,z)

* x -->> character to delete
* y -->> character to replace with
* z -->> character to delete

```python
string.punctuation
```

```python
#Tokenize and filter stopwords

filtered_tokens = [w for w in clean_phrase.lower().split() if w not in en_stopwords]

" ".join(filtered_tokens)
```

```python
texts = ["The weather today is exceptionally pleasant, beautiful, calm and enjoyable",
         "I have always enjoyed learning new things that is why I am enjoying AI",
         "My upcoming course will be on Azure Cloud"]


op = []
for l in texts:
  no_punctuation_line = l.translate(translator).lower()
  op.append(
      " ".join([t for t in no_punctuation_line.split() if t not in en_stopwords])
      )



print(op)
```
