# pre-processing

#### Lowercase

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data['review_lowercase_nopunct'] = data.apply(lambda x : re.sub(r"([^\w\s])","",x['review_lowercase']),axis=1)
```

```python
data['review_lowercase_nopunct_nostopwords'] = data['review_lowercase_nopunct'].apply(lambda x : ' '.join([w for w in x.lower().replace("'",'').split() if w not in en_stopwords]))
```

```python
data
```

# Tokenizing

```python
data['tokenized'] = data.apply(lambda x : word_tokenize(x['review_lowercase_nopunct_nostopwords']),axis=1)
```

```python
data.head()
```

```python
data['tokenized'][1004]
```

# Stemming

```python
ps = PorterStemmer()
```

```python
data['stemmed'] = data['tokenized'].apply(lambda tokens : [ps.stem(t) for t in tokens])
```

```python
data.head()
```

# Lemmatization

```python
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
data['lemmatized'] = data['tokenized'].apply(lambda tokens : [lemmatizer.lemmatize(t) for t in tokens])

```

```python
data['lemmatized'][111]
```

N-Grams

```python
tokens_clean = sum(data['lemmatized'],[])
```

```python
#Unigrams

unigrams = pd.Series(nltk.ngrams(tokens_clean,1)).value_counts()
print(unigrams)
```

```python
#Bigrams

Bigrams = pd.Series(nltk.ngrams(tokens_clean,2)).value_counts()
print(Bigrams)
```

```python
#Tri-grams

Bigrams = pd.Series(nltk.ngrams(tokens_clean,3)).value_counts()
print(Bigrams)
```

```python
#N-grams

Bigrams = pd.Series(nltk.ngrams(tokens_clean,12)).value_counts()
print(Bigrams)
```
