```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

### Sentence Tokenization

```python
para = "Mr. Sharma went to Russia last year. He  brought gifts for his nephew. It was their first trip abroad !!"
```

```python
sent_tokenize(para)
```

# Word Tokenization

```python
line  = "Riya congratulated Raj on his new venture."


word_tokenize(line)
```

```python
from string import punctuation
import re
```

```python
text = "Hello everyone! How are you doing today? I am learning Tokenization."

tokens = word_tokenize(text)

cleaned_tokens = [w for w in tokens if w not in punctuation]
cleaned_tokens
```

```python
re.sub(r'[^\w\s]','',' '.join(cleaned_tokens))
```
