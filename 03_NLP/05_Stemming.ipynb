* connected, connecting, connection, connectedness -->> connect

* studying, studied -->> study, studi

```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer
```

```python
words = ['played','player','happiness','studies','cats','bats','better']
```

```python
ps = PorterStemmer()
```

```python
for w in words:
  print(w," : ",ps.stem(w))
```

```python
new_tokens = ['connected', 'connecting', 'connection', 'connectedness', 'connects','connector']
```

```python
for t in new_tokens:
  print(t," : ",ps.stem(t))
```

```python
like_tokens = ['likes','better','worse']
```

```python
for t in like_tokens:
  print(t," : ",ps.stem(t))
```

```python
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()
```

```python
print(f"{'Word':<15}{'Porter':<15}{'Snowball':<15}{'Lancaster':<15}")
print("-"*55)

for w in new_tokens:
  print(f"{w:<15}{ps.stem(w):<15}{snowball.stem(w):<15}{lancaster.stem(w):<15}")
```

```python
sentence = "The Argentina players were happily celebrating their victory after winning their match against England"
```

```python
new_tokens = sentence.lower().split()
```

```python
print(f"{'Word':<15}{'Porter':<15}{'Snowball':<15}{'Lancaster':<15}")
print("-"*55)

for w in new_tokens:
  print(f"{w:<15}{ps.stem(w):<15}{snowball.stem(w):<15}{lancaster.stem(w):<15}")
```

* Stemming : Rule-based, fast, can produce non-words
* Lemmatization : Dictionary -based, slower, produces accurate and valid words
