```python
connect_tokens = ['connected', 'connecting', 'connection', 'connectedness', 'connects','connector']

like_tokens = ['likes','better','worse']

learn_tokens = ['learned','learning','learn','learnt','learns','learner','learners']
```

```python
import nltk
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer, PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
for t in connect_tokens:
  print(t, " : ", ps.stem(t))
```

```python
for t in like_tokens:
  print(t, " : ", ps.stem(t))
```

```python
for t in learn_tokens:
  print(t, " : ", ps.stem(t))
```

```python
#Lemmatization
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
  print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in learn_tokens:
  print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in like_tokens:
  print(t, " : ", lemmatizer.lemmatize(t))
```

```python
words = ['played','player','happiness','studies','cats','bats','better']
```

```python
for t in words:
  print(t, " : ", lemmatizer.lemmatize(t))
```

Stemming vs lemmatization
* Rule based vs dictionary+linguistic (vocab + grammar)
* Accuracy : Fast and crude vs Slow but accurate
* Result : Often not a real word vs Always a valid word
* POS : No vs Yes
* Use case : Large-scale op like search engine or indexing vs Meaning-preserving NLP (chatbots, text analysis, etc.)
