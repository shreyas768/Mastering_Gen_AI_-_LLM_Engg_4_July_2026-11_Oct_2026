```python
headline_df['lower_text'] = headline_df['title'].str.lower()
```

```python
nltk.download('stopwords')
```

```python
#remove punctuation

headline_df['no_punct'] = headline_df.apply(lambda x : re.sub(r"([^\w\s])","",x['lower_text']),axis=1)
```

```python
#Remove stopwords
stop_words = stopwords.words('english')
headline_df['cleaned_title'] = headline_df['no_punct'].apply(
lambda sentence : ' '.join([w for w in sentence.split() if w not in stop_words])
)
```

```python
headline_df
```

```python
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
headline_df['raw_tokens'] = headline_df.apply(lambda row : word_tokenize(row['cleaned_title']),axis=1)
lem = WordNetLemmatizer()
headline_df['lemm_tokens'] = headline_df['raw_tokens'].apply(
    lambda token_list : [lem.lemmatize(t) for t in token_list]
)
```

```python
headline_df.head()
```

```python
cleaned_token_list = sum(headline_df['lemm_tokens'],[])
```

# POS tagging

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(' '.join(cleaned_token_list))
```

```python
#extract tokens and pos tags

pos_tag_df = pd.DataFrame(columns = ['token','pos_tag'])
```

```python
for t in doc:
  pos_tag_df = pd.concat([
      pos_tag_df,
      pd.DataFrame.from_records([{
          'token':t.text,
          'pos_tag':t.pos_
      }])
  ],ignore_index=True)

pos_tag_df
```

```python
pos_counts = pos_tag_df.groupby(['token','pos_tag']).size().reset_index(name='counts').sort_values(by='counts',ascending=False)
pos_counts.head()
```
