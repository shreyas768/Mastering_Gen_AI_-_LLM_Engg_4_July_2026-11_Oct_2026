```python
text = """
Data Science is an interdisciplinary field that uses scientific and statistical methods,
processes and algorithms to extract insights from data. Data Scientists work with
structured and unstructured data to uncover hidden patterns and predict outcomes. In coming years,
Data Science will be essential for companied trying to gain an edge over others.
"""
```

```python
tokens = nltk.word_tokenize(text.lower())
```

```python
tokens #3/4 of a word
```

```python
#Unigrams (N=1)

unigrams = list(nltk.ngrams(tokens,1))
```

```python
unigrams = pd.Series(nltk.ngrams(tokens,1)).value_counts()
```

```python
unigrams[:10].sort_values().plot.barh(color='red')
```

```python
#Bigrams (N=2)

list(nltk.ngrams(tokens,2))
```

```python
bigrams = pd.Series(nltk.ngrams(tokens,2)).value_counts()
```

```python
bigrams[:10].sort_values().plot.barh(color='red')
```

```python
#Trigrams (N=3)

tuple(nltk.ngrams(tokens,3))
```
