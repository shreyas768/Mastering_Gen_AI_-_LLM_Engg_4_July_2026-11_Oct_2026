# Regex Syntac


* . matches any character except a new line
* ^ matches the start of the string
* $ matches the end of the string
* * matches zero or more repetitions
* ? makes the preceding character optimal
* | "OR" operators

```python
feedback = [

            "He was so helpful during my visit to Sydney",
            "Customer Support was very prompt.",
            "Great work !!!",
            "Garry ensured I reach there on time.",
            "I have pen, pencil, paper and everything else you want !!",
            "Great experience, all thanks to Jarry"
]
```

```python
# Remove Punctuation from feedback

clean_feedback = []
remove_punct = r"[^\w\s]" #remove non-word, non-space characters
for t in feedback:
  clean_feedback.append(re.sub(remove_punct,"",t))

print(clean_feedback)
```

```python
feedback
```

```python
#Find comments related to Garry (regardless of the  spelling)

spelling_with_Garry = []

for text in feedback:
  if re.search(r"j?arry", text, re.IGNORECASE):
    spelling_with_Garry.append(text)

print(spelling_with_Garry)
```

```python
#Find comments containing 'was' and 'to'

key_comments = []
items = r"was|to|work"
for text in feedback:
  if re.search(items, text, re.IGNORECASE):
    key_comments.append(text)

print(key_comments)
```

```python
#feeback ending with 'y'

op = []
end_y = r"y$"

for comment in feedback:
  if re.search(end_y,comment, re.IGNORECASE):
    op.append(comment)

print(op)
```

```python
#feeback startign with 'G

op = []
end_y = r"^Great"

for comment in feedback:
  if re.search(end_y,comment, re.IGNORECASE):
    op.append(comment)

print(op)
```
