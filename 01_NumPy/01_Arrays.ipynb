# Built-In Methods to create Arrays

```python
#arange : return evenly spaced values within the range provided

np.arange(0,10)
```

```python
np.arange(101,118) #default step size is 1
```

```python
np.arange(101,98,-1)
```

```python
np.arange(101,1118,90)
```

```python
np.arange(6)
```

```python
np.arange(16,3)
```

```python
# zeros and ones

np.zeros(4)
```

```python
np.zeros((3,5))
```

```python
np.ones(4)
```

```python
np.ones((3,5))
```

```python
#linspace : returns evenly spaces numbers over a specified intervals

np.linspace(0,10,5)
```

```python
np.linspace(0,10,85)
```

```python
#eye
np.eye(5)
```

```python
#randint : return random integers from low to high

arr1 = np.random.randint(80,99,100)

arr1
```

```python
arr1.reshape(50,2)
```

```python
#seed : to set random state so that same random results can be reproduced regardless of machine, time, system etc.

np.random.seed(67)
np.random.randint(80,99,100)
```

```python
np.random.seed(67)
np.random.randn(5,3)
```
