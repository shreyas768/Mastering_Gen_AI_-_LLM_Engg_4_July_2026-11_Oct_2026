# Process Breakdown

```python
bert_model = "google-bert/bert-base-uncased"

#bert_model = "xlnet/xlnet-base-cased"

bert_tokenizer = AutoTokenizer.from_pretrained(bert_model)
```

```python
example_sentence = "Excited to learn language models today"

encoded = bert_tokenizer(example_sentence)
print("Encoded - BERT : ",encoded)


bert_tokens = bert_tokenizer.tokenize(example_sentence)
print(f"Tokens : {bert_tokens}")
```

```python
bert_token_ids = bert_tokenizer.convert_tokens_to_ids(bert_tokens)
print("Token IDs (BERT) : ",bert_token_ids)

decoded_from_ids = bert_tokenizer.decode(bert_token_ids)
print("Decoded back (BERT) : ", decoded_from_ids)
```

# Hugging Face and Pytorch/Tensorflow

```python
example_sentence = "I hate talking to him"
example_sentence
```

```python
encoded =bert_tokenizer(example_sentence)
encoded
```

```python
distil_tokenizer = AutoTokenizer.from_pretrained('distilbert/distilbert-base-uncased-finetuned-sst-2-english')

inputs_pt = distil_tokenizer(example_sentence,return_tensors='pt')

print("Input for Pytorch Model: ", inputs_pt)
```

```python
clf_model = AutoModelForSequenceClassification.from_pretrained('distilbert/distilbert-base-uncased-finetuned-sst-2-english')

with torch.no_grad():
  out = clf_model(**inputs_pt).logits
```

```python
out
```

```python
pred_id = out.argmax().item()


print("Predicted label name: ", clf_model.config.id2label[pred_id])
```
