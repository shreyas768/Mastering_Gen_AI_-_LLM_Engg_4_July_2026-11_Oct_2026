# Customizing the Output


* Temperature : controls how random the model can get.
* 0 : very pre dictable
* 1 : more create and unpredictable

```python
def make_text(prompt, max_tokens = 50, temp = 0.5):
  resp = client1.completions.create(
      model = "gpt-3.5-turbo-instruct",
      prompt=prompt,
      max_tokens = max_tokens,
      temperature=temp
  )
  return resp.choices[0].text.strip()
```

```python
text = "In India, it is very common to"

print(f"Low Temp Result :\n{make_text(prompt=text,temp=0)}")
print(f"\n\nHigh Temp Result :\n{make_text(prompt=text,temp=1)}")
```

* System Message : Set the ground rules, instructions that tell the model how to behave and what its job is.
* User Message : Represent what the user would type
* Assistant Message : These are the egs of how the output should be.

```python
def extract_keywords(sample_text):
  response = client1.chat.completions.create(
      model = "gpt-3.5-turbo",
      messages = [
          {"role":"system","content":"You are an assistant who specializes in pulling out important keywords from text. Your response should be concise in a comma-separated list of extracted keywords, with no other explanation."},
          {"role":"user","content":"Hi, I have a paragraph that I want you to have a look. Before i send it, can you tell me what exactly you extract?"},
          {"role":"assistant","content":"Sure. Iextract the main ideas, titles, entities, key phrases and important descriptors"},
          {"role":"user","content":"Great ! Let me give a simple paragrapht first. Please show me hwo you'd extract keywords from it"},
          {"role":"assistant","content":"Sure, please share the sample paragraph"},
          {"role":"user","content":"""At Tata Motors, we explore new possibilities and push boundaries. Our human-centric,

hi-tech approach is evident in our extensive range of integrated and smart e-mobility

solutions. We choose to stay ahead of the curve. With 'responsibility' being one of our

core values, we are committed to making responsible choices, creating a positive legacy and connecting aspirations."""},
          {"role":"assistant","content":'Tata Motors, human-electric, hi-tech, e-mobility, legacy'},
          {'role':'user','content':sample_text}

      ],
      temperature=1,
      max_tokens=200,
  )
  return response.choices[0].message.content.strip()
```

# Poetic ChatBot

```python
def poet_bot(sample_text):
  response = client1.chat.completions.create(
      model = "gpt-3.5-turbo",
      messages = [
          {"role":"system","content":"You are poet that reads a text and write a good poem based on that text."},
          {"role":"user","content":"When did the first steam engine appear"},
          {"role":"assistant","content":"A puff of gas, the turn of iron wheel, in workshops past where fires burn"},
          {'role':'user','content':sample_text}

      ],
      temperature=0.8,
      max_tokens=300,
  )
  return response.choices[0].message.content.strip()
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationalRetrievalChain
```

```python
source_url = "https://tih.iitr.ac.in/training-courseDetails/103"
```

```python
from re import split
loader = WebBaseLoader(source_url)
raw_docs = loader.load()

splitter = RecursiveCharacterTextSplitter()
chunked_docs = splitter.split_documents(raw_docs)

emb = OpenAIEmbeddings(openai_api_key=api_key)
vector_index = FAISS.from_documents(chunked_docs,emb)
```

```python
chat_memory = ConversationBufferMemory(memory_key="chat_history",return_messages=True)

qa_chain = ConversationalRetrievalChain.from_llm(
    ChatOpenAI(openai_api_key=api_key,model="gpt-3.5-turbo",temperature=0),
    vector_index.as_retriever(),
    memory=chat_memory
)
```

```python
user_query = input("Type you query about this course: ")
answer = qa_chain({'question':user_query})

print(answer['answer'])
```

# Interactive Q&A

```python
print("Start asking Qs about this course. Type 'quit' to exit\n\n")

while True:
  user_query = input("Type you query about this course: ")
  if user_query.lower() == 'quit':
    print("Thanks for using this Q&A")
    break

  if user_query:
    response = qa_chain({'question':user_query})
    print("Answer:",response['answer'])
    print("\n\n")
  else:
    print("Please enter a Question\n\n")
```
